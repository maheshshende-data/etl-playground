````python
#installing requied libraries
!pip install pandas

#import requird libraries
import pandas as pd
from pyspark.sql import SparkSession

#Reading data from source(github- Customer)
df=pd.read_csv("https://raw.githubusercontent.com/Bhevendra/ML-Datasets/refs/heads/main/retail_data/sales.csv")

#creating spark session
spark = SparkSession.builder.getOrCreate()

#creating pandas dataframe to spark dataframe
df=spark.createDataFrame(df)
````

````python
from pyspark.sql.functions import col, from_json, schema_of_json

# sample one JSON string
sample_json = df.select("product").filter(col("product").isNotNull()).first()[0]

# infer schema
json_schema = schema_of_json(sample_json)

# flatten
df_flat = df.withColumn("product_json", from_json(col("product"), json_schema)) \
    .select("*", "product_json.*") \
    .drop("product", "product_json")

display(df_flat)
````

````python
df_flat.write.format("delta").mode("overwrite").saveAsTable("batch_1.data.sales")
````

### Task 2
- Fact vs Dimension
        # Fact : It represent specific business process or event like sales, login or tempmrature reading, most of the time fact are in numbers that can arranged(sum/average)
        they always answer How much?  How many? Fact always found in Large Tables, They grow very quickly

        # Dimension: dimension table shows descriptive context. represent entities like product name, location, date, etc
                    mostly use for filtering, grouping, dimensin give answer for who? when? where? what? 



- Normalization vs Denormalization

    # Normalization: the process of organising data to reduce redundancy and to improve data integrity. 

                    store data in seperate table / avoid dupication data / use relationship to connect table( foreing keys)/ better data consistancy
                    require JOINs - which slow down queries
    # Denormalization : the process of combining table to improve read performance, even if it introduce redundacy.

                        store data together/ reduce need for JOIN/ accept some duplication
                        faster read query/ smpler query/ 
                        data redundancy/ risk of inconsistency 

- Slowly Changing Dimensions (SCD)
        # it store current data and historical data, will not change very often(countiries example)
- Keys
        -# PRIMARY KEY- Unique identity/ value of Table, it never Null, there can be only one oprimary key in a Table.#
    
        -#FOREIGN KEY- the feild in one table that refere the primary key in another Table. link between two tables. can accept duplicate value.

       - #COMPOSITE KEY- consist of two or more coloumn that work together as uniquely identify record. it use when single column not enough to identify uniqueness.

        - # SURROUGATE KEY - it is an artificial identifier often generate by system, useful when natural primary key not available
